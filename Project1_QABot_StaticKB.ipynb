```python
from transformers import pipeline
```

```python
knowledge_base = """
LangChain is a powerful framework for building applications that utilize large language models (LLMs) to perform various tasks. 
It provides a structured way to create conversational agents, automate workflows, and integrate LLMs into different applications.
"""

# Load a model 
qa_pipeline = pipeline("question-answering", model='distilbert-base-cased-distilled-squad')
```

```python
def answer_question(question, context = knowledge_base):
    result = qa_pipeline(question = question, context = context)
    return result['answer'], result['score']
```

```python
question = "What is LangChain?"

answer, confidence = answer_question(question)

print(f"Question: {question}")
print(f"Answer: {answer}")
print(f"Confidence Score: {(confidence * 100):.2f}%")
```

```python
def main():
    print("Welcome to Simple QnA!")
    while True:
        question = input("Ask a question (or type 'exit' to quit): ")
        if question.lower() == "exit":
            print("Goodbye!")
            break

        answer, confidence = answer_question(question)
        print(f"Answer: {answer}")
        print(f"Confidence Score: {(confidence * 100):.2f}%")
```

```python
main()
```
