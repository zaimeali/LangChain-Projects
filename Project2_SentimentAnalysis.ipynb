```python
import transformers
from transformers import pipeline
transformers.__version__
```

```python
# Load the sentiment analysis pipeline
sentiment_analysis_pipeline = pipeline(
    "sentiment-analysis",
    model = "distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
def analyze_sentiment(text):
    result = sentiment_analysis_pipeline(text)
    
    sentiment = result[0]['label']
    score = result[0]['score']
    return sentiment, score
```

```python
def main():
    print("Welcome to Sentiment Analysis Bot!")
    while True:
        text = input("Enter text or 'exit' to cancel: ")
        if text == 'exit':
            print('Goodbye!')
            break
        response, confidence = analyze_sentiment(text)

        print(f"Text: {text}")
        print(f"Sentiment: {response}")
        print(f"Confidence Score: {(confidence * 100):.2f}%")
        print()
```

```python
main()
```
