```python
# Load the summarization pipeline
summarization_pipeline = pipeline(
    'text-generation',
    model="facebook/bart-large-cnn"
)
```

```python
def summarize_text(text, min_len = 30, max_len = 110):
    summary = summarization_pipeline(
        text,
        min_length = min_len,
        max_length = max_len,
        do_sample = False,
        max_new_tokens = 60
    )

    return summary[0]["generated_text"]
```

```python
text = """
Artificial Intelligence (AI) is a branch of computer science that aims to create machines capable of intelligent behavior. 
It encompasses various subfields, including machine learning, natural language processing, computer vision, and robotics. 
Machine learning, a core component of AI, allows systems to learn from data without being explicitly programmed. 
Deep learning, a subset of machine learning, uses neural networks with many layers to model complex patterns. 
AI applications are widespread, from virtual assistants like Siri and Alexa to self-driving cars and medical diagnosis tools. 
However, AI also raises ethical concerns, such as job displacement, privacy issues, and the potential for biased decision-making. 
As AI continues to evolve, researchers are working on making it more transparent, fair, and aligned with human values. 
The future of AI holds promise for solving global challenges like climate change and healthcare, 
but it requires careful regulation to ensure beneficial outcomes.
"""

response = summarize_text("Summarize the text: " + (text))
print(response)
```
