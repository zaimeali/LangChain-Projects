```python
from transformers import pipeline
```

```python
# Load the model
keyword_pipeline = pipeline(
    'ner',
    model = "ml6team/keyphrase-extraction-distilbert-inspec"
)
```

```python
def extract_keywords(text):
    keywords = keyword_pipeline(text)
    extracted_keywords = [keyword['word'] for keyword in keywords]
    return extracted_keywords
```

```python
extract_keywords("What a beautiful day today!. I would go out and do grocery and after it I will go to Gym")
```

```python
def main():
    print("Welcome to Keyword Extractor!")
    while True:
        text = input('Enter text or exit: ')
        if text == 'exit':
            print('Goodbye!')
            break
        keywords = extract_keywords(text)
        print("Extracted Keywords are: ", keywords)

```

```python
main()
```
